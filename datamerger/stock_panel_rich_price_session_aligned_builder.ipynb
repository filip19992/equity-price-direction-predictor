# Rich-Price Full-History Session-Aligned Panel Builder

Builds a new dataset that uses rich OHLC stock inputs from `stock_price_rich` and leaves the older datasets untouched.

Expected rich stock source files are named like `stock-prices-rich-data_aapl.csv` and optional backfill files like `stock-prices-rich-data_aapl_20210101_20221231.csv`.

The resulting raw panel uses `close_stock_price` instead of the old `stock_price` column and includes adjusted open/high/low/close plus compact same-day price features.

```python
from __future__ import annotations

from pathlib import Path
import sys

import pandas as pd

PROJECT_ROOT = Path.cwd().resolve()
if not (PROJECT_ROOT / "data" / "equity_data").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

pd.set_option("display.max_columns", 220)
pd.set_option("display.width", 260)

from datamerger.rich_price_session_aligned_full_history_builder import (
    DEFAULT_BACKFILL_SUFFIXES,
    DEFAULT_TICKERS,
    NINE_TICKERS,
    build_rich_price_full_history_dataset,
)
```

```python
# Default: original nine-ticker universe, excluding AVGO.
TICKERS = NINE_TICKERS

# Use this instead if you want all 10 imported tickers, including AVGO:
# TICKERS = DEFAULT_TICKERS

BACKFILL_SUFFIXES = DEFAULT_BACKFILL_SUFFIXES
OUTPUT_PREFIX = "stock_panel_nine_tickers_session_aligned_full_history_rich_price_adjusted_google_score"

TICKERS, BACKFILL_SUFFIXES, OUTPUT_PREFIX
```

```python
result = build_rich_price_full_history_dataset(
    project_root=PROJECT_ROOT,
    tickers=TICKERS,
    backfill_suffixes=BACKFILL_SUFFIXES,
    output_prefix=OUTPUT_PREFIX,
)

panel = result["panel"]
summary = result["summary"]
source_audit = result["source_audit"]
reddit_coverage_audit = result["reddit_coverage_audit"]

print(f"Saved panel: {result['output_dataset_path']}")
print(f"Saved summary: {result['output_summary_path']}")
print(f"Saved metadata: {result['output_metadata_path']}")

summary
```
