# QQQ Market-Context Session-Aligned Panel Builder

Builds a new full-history session-aligned dataset from the original adjusted Google score panel and the saved QQQ benchmark file.

This dataset intentionally does not use the rich OHLC price fields from the previous experiment. It keeps the original adjusted close price as `close_stock_price`, keeps stock volume and alternative-data fields, and adds compact QQQ market-context features such as QQQ returns, QQQ volatility, market-relative stock returns, rolling beta to QQQ, and residual return versus QQQ.

```python
from __future__ import annotations

from pathlib import Path
import sys

import pandas as pd

PROJECT_ROOT = Path.cwd().resolve()
if not (PROJECT_ROOT / "data" / "datasets").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

pd.set_option("display.max_columns", 220)
pd.set_option("display.width", 260)

from datamerger.qqq_market_context_session_aligned_builder import (
    DEFAULT_OUTPUT_PREFIX,
    QQQ_MARKET_CONTEXT_COLUMNS,
    build_qqq_market_context_dataset,
    default_base_dataset_path,
    default_benchmark_path,
)

PROJECT_ROOT
```

```python
BASE_DATASET_PATH = default_base_dataset_path(PROJECT_ROOT)
BENCHMARK_PATH = default_benchmark_path(PROJECT_ROOT)
OUTPUT_PREFIX = DEFAULT_OUTPUT_PREFIX

BASE_DATASET_PATH, BENCHMARK_PATH, OUTPUT_PREFIX
```

```python
result = build_qqq_market_context_dataset(
    project_root=PROJECT_ROOT,
    base_dataset_path=BASE_DATASET_PATH,
    benchmark_path=BENCHMARK_PATH,
    output_prefix=OUTPUT_PREFIX,
)

panel = result["panel"]
summary = result["summary"]
source_audit = result["source_audit"]

print(f"Saved panel: {result['output_dataset_path']}")
print(f"Saved summary: {result['output_summary_path']}")
print(f"Saved source audit: {result['output_source_audit_path']}")
print(f"Saved metadata: {result['output_metadata_path']}")

summary
```
