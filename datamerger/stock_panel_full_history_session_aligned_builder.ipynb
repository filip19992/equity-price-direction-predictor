# Full-History Session-Aligned Panel Builder

Builds one raw panel from the current source files plus the backfill files tagged `20210101_20221231`.

Default output uses the same nine-ticker universe as `stock_panel_nine_tickers_session_aligned_raw.csv` and writes new files under `data/datasets/` without touching the existing dataset. AVGO can be enabled in the config cell.

```python
from __future__ import annotations

from pathlib import Path
import sys

import pandas as pd

PROJECT_ROOT = Path.cwd().resolve()
if not (PROJECT_ROOT / "data" / "equity_data").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

pd.set_option("display.max_columns", 200)
pd.set_option("display.width", 220)

from datamerger.session_aligned_full_history_builder import (
    DEFAULT_BACKFILL_SUFFIXES,
    DEFAULT_TICKERS,
    NINE_TICKERS,
    build_full_history_dataset,
)
```

```python
# Default: original nine-ticker universe, excluding AVGO.
TICKERS = NINE_TICKERS

# Use this instead if you want all 10 imported tickers, including AVGO:
# TICKERS = DEFAULT_TICKERS

BACKFILL_SUFFIXES = DEFAULT_BACKFILL_SUFFIXES
OUTPUT_PREFIX = None  # auto: stock_panel_nine_tickers_session_aligned_full_history

TICKERS, BACKFILL_SUFFIXES
```

```python
result = build_full_history_dataset(
    project_root=PROJECT_ROOT,
    tickers=TICKERS,
    backfill_suffixes=BACKFILL_SUFFIXES,
    output_prefix=OUTPUT_PREFIX,
)

print(f"Saved panel to {result['output_dataset_path']}")
print(f"Saved summary to {result['output_summary_path']}")
print(f"Saved metadata to {result['output_metadata_path']}")
print(f"Saved source audit to {result['output_source_audit_path']}")
print(f"Saved reddit coverage audit to {result['output_reddit_coverage_path']}")
```
